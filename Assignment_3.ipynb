```python
def right_triangle(c,a,b):
    if c**2 == a**2 + b**2:
        return "Right Angled triangle"
    else:
        return "Not a Right Angled triangle"
print(right_triangle(5,3,4))
print(right_triangle(5,2,4))
print(right_triangle(int(input("Enter the length of the first side: ")),int(input("Enter the length of the second side: ")),int(input("Enter the length of the third side: "))))

```

```python
#using sort function
x= int(input("Enter side x: "))
y= int(input("Enter side y: "))
z= int(input("Enter side z: "))
def right_triangle(x,y,z):
    sides = [x, y, z]
    sides.sort()
    if sides[2]**2 == sides[0]**2 + sides[1]**2:
        return "Right Angled triangle"
    else:
        return "Not a Right Angled triangle"
print(right_triangle(x,y,z))
```
